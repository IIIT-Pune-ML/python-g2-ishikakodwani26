## Strings in Python

Open the file poem.txt and print lenght of each line.

```python
with open('poem.txt','r') as file:
   lines=file.readlines()


for i, line in enumerate(lines, 1):
    print(f"Line {i}: {len(line.strip())}")
```

Find sum of each line, mean, median and standard deviation of all lines in the file.

```python
import statistics
line_lengths = [len(line.strip()) for line in lines]

print("Sum:", sum(line_lengths))
print("Mean:", statistics.mean(line_lengths))
print("Median:", statistics.median(line_lengths))
print("Standard Deviation:", statistics.stdev(line_lengths))
```

```python
for key in x.keys():
    print(key)
```

Take a list of 10 elements and convert into json using  json library. Demonstrate use of loads and dumps

```python
import json

my_list = [10, 20, 30, 40, 50, "Apple", "Banana", True, 3.14, "Python"]

json_data = json.dumps(my_list)

print("JSON using dumps():")
print(json_data)

python_list = json.loads(json_data)

print("\nPython list using loads():")
print(python_list)

print("\nData type after loads():", type(python_list))
```
